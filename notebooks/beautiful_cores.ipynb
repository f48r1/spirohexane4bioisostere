```python
import pandas as pd
from rdkit import Chem
from rdkit.Chem import AllChem

from src.descriptors import strain_energy
from src.dataloader import special_cores, Dataloader
```

```python
dataloader = Dataloader()
dataloader.load_cores()
dataloader.load_descriptions()
dataloader.load_dft_mols()
```

```python
mols = dataloader.dft_mols.copy()
mols.index = dataloader.smiles
```

```python
descriptions = dataloader.descriptions.set_index('smiles')
descriptions
```

```python
descriptions['strain'] = dataloader.smiles.apply(strain_energy, dft_dir = 'data/cores/DFT').values
descriptions['volume'] = mols.apply(AllChem.ComputeMolVolume)
descriptions
```

```python
ref_smiles = special_cores[0]
tgt_smiles = special_cores[1:]
```

```python
delta_perc_df = pd.DataFrame()
delta_perc_df['smiles'] = tgt_smiles
for descriptor in ['PBF','volume','dipole','strain']:
	delta_perc_df[descriptor] = delta_perc_df.smiles.apply(
		lambda x: (descriptions.loc[x,descriptor] - descriptions.loc[ref_smiles,descriptor]) * 100 / descriptions.loc[ref_smiles,descriptor]
	)
```

```python
delta_perc_df
```

```python
delta_df = pd.DataFrame()
delta_df['smiles'] = tgt_smiles
for descriptor in ['PBF','volume','dipole','strain']:
	delta_df[descriptor] = delta_df.smiles.apply(
		lambda x: descriptions.loc[x,descriptor] - descriptions.loc[ref_smiles,descriptor]
	)
```

```python
delta_df
```
